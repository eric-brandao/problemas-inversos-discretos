# 1. Introdução e motivação

Os modelos matemáticos e os dados desempenham um papel importante na ciência e na engenharia modernas. O campo dos problemas inversos faz uma ponte entre esses dois domínios e estuda se e como é possível **inferir parâmetros** de modelos a partir de **medições**, que são frequentemente **incompletas**, **esparsas** e/ou **contaminadas por ruído**. 

Podemos falar de forma superficial sobre alguns exemplos de problemas inversos. 

### Ex 1. Ultrassonografia médica.

Uma ultrassonografia médica é um exame de imagem não invasivo. A ideia é a seguinte:

1. Um emissor é colocado sobre a pele do paciente e emite uma onda sonora de alta frequência ($\approx [MHz]$). Esta onda vai se propagar para o interior do corpo do paciente e ser difratada (refletida) de volta à origem.
2. Tipos especiais de microfones medem a onda sonora difratada pelo corpo do paciente. Agora temos a primeira parte do quebra-cabeça: o **dado medido**. Note que o dado medido será contaminado por ruído (*e.g.*, elétrico, acústico, paciente se mexe, etc.). Note também que o sensor capta o som difratado numa região limitada do espaço - portanto, o dado medido é incompleto.
3. Um **modelo matemático** é usado pra inferir como o som se propaga pelo interior do corpo humano e retorna à origem. Os **parâmetros** deste modelo precisam ser determinados de alguma forma — essa é a tarefa de uma formulação de problema inverso. O modelo matemático é a segunda parte do quebra-cabeça.
4. Uma vez que se tem uma boa estimativa dos parâmetros do modelo matemático, conseguimos **reconstruir** a onda sonora em locais onde não medimos (no interior do corpo humano). Assim, conseguimos imagens estimadas do interior do corpo humano.

### Ex 2. Uma imagem icônica.

A imagem icônica do buraco negro na Figura 2 foi obtida da seguinte maneira:

1. Uma coleção de telescópios na superfície da Terra mediu o campo elétrico emitido na direção do buraco negro em função do tempo, da frequência e da polarização. Isto compreende o **dado medido**, contaminado por ruído, incompleto e esparso. 
2. Um **modelo matemático** é usado pra inferir como a radiação eletromagnética se propaga pelo espaço-tempo. Os **parâmetros** deste modelo foram determinados de alguma forma. 
3. Com a estimativa dos parâmetros do modelo matemático, conseguimos **reconstruir** a imagem do emissor (no caso, o buraco negro).

# Em suma

Nós vamos precisar entender o seguinte:

1. A natureza dos dados à nossa disposição.
2. O modelo matemático.
3. Como encontrar os **parâmetros** de maneira sensata.
4. Reconstruir é, em geral, uma tarefa trivial. Ao menos, comparado ao esforço inicial.

Logo, os problemas inversos são alguns dos problemas matemáticos mais importantes nas ciências e na matemática, pois nos informam sobre parâmetros que não podemos observar diretamente. Um esquema da nossa tarefa é apresentado a seguir: 

- nós conhecemos as saídas (medidas com algum erro ou incerteza);
- também conhecemos (em geral) o modelo matemático (ou sistema);
- o que desejamos **inferir** são as entradas (a causa por trás do que medimos);

# O básico de um problema inverso

Vamos começar de modo geral. O modelo matemático de um problema inverso pode ser escrito da seguinte forma

\begin{equation}
d(\mathbf{r},t)= f(\mathbf{\Theta}, \mathbf{r},t) + n(\mathbf{r},t),
\tag{1}
\end{equation}
em que 

- $d(\mathbf{r},t)$ representa o dado medido em função das variáveis $\mathbf{r}$ (espacial) e $t$ (temporal);
- $f(\mathbf{\Theta}, \mathbf{r},t)$ representa o modelo matemático, que além de ser função de $\mathbf{r}$ e $t$, também é função de uma série de parâmetros $\mathbf{\Theta}$ (que é o que queremos estimar);
- $n(\mathbf{r},t)$ representa o ruído presente em toda medição, responsável por tornar nossa observação $d$ diferente do que seria perfeitamente previsto pelo modelo matemático $f$.

A Eq. (1) é genérica no sentido de que tanto os dados medidos quanto o modelo matemático podem ser **contínuos**. Além disso, o modelo matemático pode ser linear ou não-linear. No entanto, nossas medições serão sempre discretas (uma coleção de dados organizada em um **vetor**). Além disso, frequentemente, o modelo matemático pode ser **linear**. Pensando dessa forma, pode-se expressar o valor de uma única medição como a seguinte combinação linear


\begin{equation}
d(\mathbf{r}_m,t)= \sum\limits_{i=1}^{L}\tilde{x}_i \phi_i(\mathbf{r}_m,t) + n(\mathbf{r}_m,t),
\tag{2}
\end{equation}
em que $\phi_i(\mathbf{r}_m,t)$ é uma função de base adequada ao problema em mãos (p. ex., *point spread function* — para imagens; uma função de Green — para propagação ondulatória, etc.). Note que a Eq.(2) é um **produto interno**. De fato, podemos reescrevê-la da seguinte forma

\begin{equation}
d(\mathbf{r}_m,t)= \left[\phi_1(\mathbf{r}_m,t), \ \phi_2(\mathbf{r}_m,t), \ , \cdots, \ \phi_L(\mathbf{r}_m,t)\right] \ \left[x_1, \ x_2, \ \cdots, \ x_L\right]^T + n(\mathbf{r}_m,t).
\tag{3}
\end{equation}

Além disso, a Eq.(3) pode ser escrita para diversos pontos de medição, formando um sistema de equações, a saber

\begin{equation}
\underbrace{
\begin{bmatrix}
d(\mathbf{r}_1,t) \\ d(\mathbf{r}_2,t) \\ \vdots \\ d(\mathbf{r}_M,t)
\end{bmatrix}}_{\text{medições}}
=
\underbrace{\begin{bmatrix}
\phi_1(\mathbf{r}_1,t) & \phi_2(\mathbf{r}_1,t) & \cdots &\phi_L(\mathbf{r}_1,t) \\
\phi_1(\mathbf{r}_2,t) & \phi_2(\mathbf{r}_2,t) & \cdots &\phi_L(\mathbf{r}_2,t) \\
\vdots & \vdots & \ddots & \vdots \\
\phi_1(\mathbf{r}_M,t) & \phi_2(\mathbf{r}_M,t) & \cdots &\phi_L(\mathbf{r}_M,t)
\end{bmatrix}}_{\text{modelo}}
\underbrace{\begin{bmatrix}
\tilde{x}_1 \\ \tilde{x}_2 \\ \vdots \\ \tilde{x}_L
\end{bmatrix}}_{\text{incógnitas}}
\tag{4}
\end{equation}


Isso implica que a Eq. (1) pode ser transformada numa versão matricial, dada por

\begin{equation}
\mathbf{b}= \mathbf{A} \mathbf{x} + \mathbf{n},
\tag{5}
\end{equation}

em que 

- $\mathbf{b} = \left[b_1, b_2, \cdots b_M\right]^T \in \mathbb{C}^{M \times 1}$ é o vetor de observações (ou medições); ele contém uma série de números (reais ou, de forma genérica, complexos);
- $\mathbf{A} \in \mathbb{C}^{M \times L}$ é a matriz de sensibilidade do problema; ela contém uma série de números (ou coeficientes), que são computados pela função $f(\mathbf{\Theta}, \mathbf{r},t)$ medidos discretamente nas variáveis independentes;
- $\mathbf{x} = \left[x_1, x_2, \cdots x_L\right]^T \in \mathbb{C}^{L \times 1}$ é o vetor de coeficientes de $\mathbf{\Theta}$ com a solução desejada do problema inverso.
- $\mathbf{n} \in \mathbb{C}^{M \times 1}$ é o ruído associado à medição (frequentemente desconhecido).

# Sendo ingenuamente otimistas

Vamos ser otimistas e pensar que talvez nossas medições sejam feitas com tamanho cuidado que $\mathbf{n} \approx \mathbf{0}$. Neste caso, podemos tentar encontrar $\mathbf{x}$ da seguinte forma

\begin{equation}
\mathbf{\tilde{x}}= \mathbf{A}^{-1} \mathbf{b},
\tag{6}
\end{equation}
em que $\mathbf{A}^{-1}$ é a inversa de $\mathbf{A}$, que com alguma (ou muita) sorte existe. 

Vamos chamar $\mathbf{\tilde{x}}$ na Eq.(3) de **solução ingênua**. Veremos depois por que é chamada assim. 

Por ora, espera-se que uma coisa esteja clara: vetores e matrizes serão de suma importância nessa disciplina. Assim, podemos começar fazendo uma revisão sobre eles, suas principais propriedades e algumas operações importantes.
